```python
data = ['data/testing_walk1.txt', 'data/testing_fall1.txt', 'data/testing_fall2.txt']

for file in data:
    with open(file, 'r') as f:
        qpos = [line.strip().split(',') for line in f]
        qpos = [[float(x) for x in row] for row in qpos]
        
    print(f"Data from {file} is walking forward: ", is_walking_forward(qpos))
    print(f"Data from {file} has fallen: ", has_fallen(qpos))
```

```python
for i in range(6):
    plt.figure()
    for file in data:
        with open(file, 'r') as f:
            qpos = [line.strip().split(',') for line in f]
            qpos = [[float(x) for x in row] for row in qpos]
        x = [frame[i] for frame in qpos]
        plt.plot(x[:-1], label=f"{file}_{i}")
    plt.legend()
    plt.title(f"Joint {i} Position Over Time")
    plt.xlabel("Time Step")
    plt.ylabel("Position")
    plt.show()
```
